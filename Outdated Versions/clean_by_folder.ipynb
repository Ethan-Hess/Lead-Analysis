```python
import pandas as pd
from pathlib import Path
import glob
from concurrent.futures import ThreadPoolExecutor

# Define the path to the directory containing the Excel files
path = Path('D:/VSCode/Lead Analysis/Data/82M Part-2')

# List all Excel files in the directory
all_files = glob.glob(str(path / "*.xlsx"))

print(all_files)

# Function to get column names from an Excel file
def get_column_names(file):
    df = pd.read_excel(file, nrows=0)
    print(f"Columns in {file}: {df.columns}")
    return df.columns

# Use ThreadPoolExecutor to get column names in parallel
with ThreadPoolExecutor() as executor:
    futures = [executor.submit(get_column_names, file) for file in all_files]
    results = [future.result() for future in futures]

# Print the results
for file, columns in zip(all_files, results):
    print(f"Columns in {file}: {columns}")
```
